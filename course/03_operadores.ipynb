# Lección 3: Operadores y matemáticas

Python es una calculadora muy poderosa. 🧮 En la programación competitiva casi todos los problemas usan estas operaciones.

## Las operaciones básicas

| Operador | Qué hace | Ejemplo | Resultado |
|---|---|---|---|
| `+` | suma | `7 + 2` | `9` |
| `-` | resta | `7 - 2` | `5` |
| `*` | multiplicación | `7 * 2` | `14` |
| `/` | división normal | `7 / 2` | `3.5` |
| `//` | división entera | `7 // 2` | `3` |
| `%` | residuo (módulo) | `7 % 2` | `1` |
| `**` | potencia | `7 ** 2` | `49` |

```python
print(7 + 2)
print(7 - 2)
print(7 * 2)
print(7 / 2)
print(7 // 2)
print(7 % 2)
print(7 ** 2)
```

## División entera y residuo

Imagina que repartes 17 caramelos entre 5 amigos:

- Cada uno recibe `17 // 5` = **3** caramelos.
- Sobran `17 % 5` = **2** caramelos.

Estas dos operaciones aparecen **todo el tiempo** en los problemas.

```python
caramelos = 17
amigos = 5
print("Cada uno recibe", caramelos // amigos)
print("Sobran", caramelos % amigos)
```

### Truco: ¿par o impar?

Un número es **par** si al dividirlo entre 2 no sobra nada: `n % 2 == 0`.

```python
n = 14
print(n % 2)     # 0 significa par, 1 significa impar
```

### Truco: separar dígitos

El último dígito de un número es `n % 10`. Quitarlo es `n // 10`.

```python
n = 3472
print("Último dígito:", n % 10)
print("Sin el último dígito:", n // 10)
```

## Orden de las operaciones

Como en matemáticas, primero `**`, luego `*` `/` `//` `%`, y al final `+` `-`. Con paréntesis tú decides el orden.

```python
print(2 + 3 * 4)       # 14
print((2 + 3) * 4)     # 20
print(2 ** 3 ** 2)     # 512: la potencia se evalúa de derecha a izquierda
```

## Números enormes sin miedo

En muchos lenguajes los números grandes se "desbordan". En Python **no**: los enteros pueden ser tan grandes como quieras. Esto es una gran ventaja en competencias.

```python
print(2 ** 100)
print(1000000 * 1000000 * 1000000)
```

## Atajos para actualizar variables

| Escribes | Equivale a |
|---|---|
| `x += 3` | `x = x + 3` |
| `x -= 3` | `x = x - 3` |
| `x *= 3` | `x = x * 3` |
| `x //= 3` | `x = x // 3` |
| `x %= 3` | `x = x % 3` |

```python
x = 10
x += 5
print(x)
x *= 2
print(x)
```

## Funciones matemáticas útiles

```python
print(abs(-8))             # valor absoluto: 8
print(min(4, 9, 2))        # el menor: 2
print(max(4, 9, 2))        # el mayor: 9
print(round(3.7))          # redondeo: 4
print(divmod(17, 5))       # cociente y residuo juntos: (3, 2)
```
