# Lección 2: Variables y tipos de datos

Una **variable** es una caja con nombre donde guardas un valor para usarlo después. 📦

```python
edad = 12
print(edad)
```

Leído en voz alta: "`edad` **guarda** el valor 12". El signo `=` no significa "es igual" como en matemáticas: significa **guardar**.

Puedes cambiar el valor cuando quieras:

```python
puntos = 10
print(puntos)
puntos = 25
print(puntos)
puntos = puntos + 5     # toma el valor actual (25), súmale 5 y guárdalo otra vez
print(puntos)
```

### Reglas para los nombres
- Pueden tener letras, números y `_`, pero **no empezar con número**.
- No pueden tener espacios: usa `mi_edad`.
- Mayúsculas y minúsculas son distintas: `Edad` y `edad` son cajas diferentes.
- Elige nombres que expliquen qué guardan: `goles` es mejor que `x`.

## Tipos de datos

Cada valor tiene un **tipo**. Los cuatro más importantes:

| Tipo | Nombre | Ejemplo |
|---|---|---|
| `int` | número entero | `7`, `-3`, `0` |
| `float` | número con decimales | `3.14`, `-0.5` |
| `str` | texto | `"hola"` |
| `bool` | verdadero o falso | `True`, `False` |

Con `type(...)` puedes preguntarle a Python el tipo de un valor:

```python
print(type(7))
print(type(3.14))
print(type("hola"))
print(type(True))
```

## Texto y números no se mezclan solos

```python
a = "3"
b = "4"
print(a + b)       # con textos, + los pega: "34"

c = 3
d = 4
print(c + d)       # con números, + suma: 7
```

## Convertir entre tipos

- `int(x)` convierte a entero.
- `float(x)` convierte a decimal.
- `str(x)` convierte a texto.

```python
texto = "42"
numero = int(texto)
print(numero + 1)

print(str(100) + " puntos")
```

## Leer datos con `input`

`input()` espera a que el usuario escriba algo y siempre lo entrega **como texto**. Si quieres un número, conviértelo con `int(...)`.

Ejecuta la celda y escribe en el cuadro que aparece arriba.

```python
nombre = input("¿Cómo te llamas? ")
print("Hola,", nombre)
```

```python
edad = int(input("¿Cuántos años tienes? "))
print("El próximo año tendrás", edad + 1)
```

## Mezclar texto y variables: f-strings

Si pones una `f` antes de las comillas, lo que escribas entre `{ }` se reemplaza por el valor de la variable.

```python
nombre = "Ana"
edad = 11
print(f"{nombre} tiene {edad} años")
print(f"El año que viene tendrá {edad + 1}")
```
