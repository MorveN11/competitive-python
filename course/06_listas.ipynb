# Lección 6: Listas

Una variable guarda un valor. ¿Y si necesitas guardar 100 notas? Para eso existen las **listas**: una fila de casillas numeradas. 🗂️

```python
notas = [80, 95, 70, 100, 85]
print(notas)
```

## Índices: cada casilla tiene un número

Los índices **empiezan en 0**.

| Posición | 0 | 1 | 2 | 3 | 4 |
|---|---|---|---|---|---|
| Valor | 80 | 95 | 70 | 100 | 85 |

```python
notas = [80, 95, 70, 100, 85]
print(notas[0])      # primer elemento
print(notas[2])      # tercer elemento
print(notas[-1])     # -1 es el último
print(notas[-2])     # -2 es el penúltimo
print(len(notas))    # cuántos elementos hay
```

## Modificar una lista

```python
frutas = ["manzana", "pera"]
frutas.append("uva")         # agrega al final
print(frutas)

frutas[0] = "mango"          # cambia un elemento
print(frutas)

frutas.pop()                 # quita el último
print(frutas)
```

## Recorrer una lista

Hay dos formas. La primera usa directamente los valores:

```python
notas = [80, 95, 70]
for nota in notas:
    print(nota)
```

La segunda usa los índices (útil cuando necesitas saber la posición):

```python
for i in range(len(notas)):
    print(f"posición {i}: {notas[i]}")
```

## Funciones útiles con listas

```python
numeros = [5, 2, 9, 1, 7]
print(sum(numeros))       # suma de todo
print(min(numeros))       # el menor
print(max(numeros))       # el mayor
print(len(numeros))       # cantidad
print(sorted(numeros))    # copia ordenada (no cambia la original)
numeros.sort()            # ordena la lista misma
print(numeros)
numeros.reverse()         # la da vuelta
print(numeros)
print(7 in numeros)       # ¿está el 7?
```

## Rebanadas (slices)

`lista[a:b]` toma desde la posición `a` hasta **antes** de `b`.

```python
l = [10, 20, 30, 40, 50, 60]
print(l[1:4])     # [20, 30, 40]
print(l[:3])      # los 3 primeros
print(l[3:])      # desde la posición 3 hasta el final
print(l[::2])     # de 2 en 2
print(l[::-1])    # al revés
```

## Crear listas rápido

```python
ceros = [0] * 5
print(ceros)

cuadrados = [i * i for i in range(1, 6)]    # lista por comprensión
print(cuadrados)

pares = [x for x in range(20) if x % 2 == 0]
print(pares)
```

Convertir texto con números en una lista de enteros. **Este patrón lo usarás en todos los problemas**:

```python
linea = "3 8 1 6"
numeros = [int(x) for x in linea.split()]
print(numeros)
```

## Matrices: listas de listas

Una tabla (filas y columnas) es una lista donde cada elemento es otra lista.

```python
matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9],
]
print(matriz[0])        # la primera fila
print(matriz[1][2])     # fila 1, columna 2 -> 6

for fila in matriz:
    print(fila)
```

Crear una matriz vacía de `n` filas y `m` columnas. ⚠️ **No uses `[[0] * m] * n`**: copiaría la misma fila varias veces. Usa esto:

```python
n, m = 3, 4
tabla = [[0] * m for _ in range(n)]
tabla[1][2] = 9
for fila in tabla:
    print(fila)
```
