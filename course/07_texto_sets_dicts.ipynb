# Lección 7: Textos, tuplas, conjuntos y diccionarios

Cuatro herramientas más para guardar y manejar datos. Las usarás todo el tiempo.

## Textos (strings)

Un texto funciona casi como una lista de letras.

```python
palabra = "python"
print(len(palabra))      # 6
print(palabra[0])        # p
print(palabra[-1])       # n
print(palabra[1:4])      # yth
print(palabra[::-1])     # nohtyp (al revés)

for letra in palabra:
    print(letra)
```

⚠️ Los textos **no se pueden modificar** letra por letra; se crea uno nuevo.

### Métodos útiles

```python
frase = "  Hola Mundo  "
print(frase.strip())            # quita espacios de los lados
print(frase.upper())            # MAYÚSCULAS
print(frase.lower())            # minúsculas
print("a b c".split())          # ['a', 'b', 'c']
print("1,2,3".split(","))       # separa por comas
print("-".join(["a", "b", "c"]))  # a-b-c
print("hola".count("o"))        # 1
print("hola".replace("o", "0")) # h0la
print("ana" == "ana"[::-1])     # ¿es palíndromo? True
```

## Tuplas

Como una lista, pero **no se puede cambiar**. Se usan para agrupar pocos datos, como una coordenada.

```python
punto = (3, 4)
print(punto[0], punto[1])

x, y = punto             # desempaquetar
print(x, y)

a, b = 1, 2
a, b = b, a              # intercambiar dos variables
print(a, b)
```

## Conjuntos (`set`)

Una bolsa de elementos **sin repetidos** y donde preguntar "¿está este elemento?" es muy rápido. ⚡

```python
s = {3, 1, 4, 1, 5, 9, 2, 6, 5}
print(s)                 # los repetidos desaparecen

s.add(10)
s.discard(3)             # quita si existe, sin error
print(s)
print(4 in s)            # True
print(len(s))
```

Quitar repetidos de una lista:

```python
datos = [5, 3, 5, 1, 3, 3, 7]
print(len(set(datos)))           # cuántos valores distintos hay
print(sorted(set(datos)))        # ordenados y sin repetidos
```

Operaciones entre conjuntos:

```python
A = {1, 2, 3, 4}
B = {3, 4, 5}
print(A & B)    # intersección: en ambos
print(A | B)    # unión: en alguno
print(A - B)    # en A pero no en B
```

## Diccionarios (`dict`)

Guardan pares **clave → valor**, como un diccionario real: buscas la palabra (clave) y obtienes el significado (valor). 📖

```python
edades = {"Ana": 11, "Luis": 12}
print(edades["Ana"])

edades["Marta"] = 10            # agrega
edades["Ana"] = 12              # cambia
print(edades)
print("Luis" in edades)         # ¿existe la clave?
print(edades.get("Pedro", 0))   # valor por defecto si no existe

for nombre in edades:
    print(nombre, edades[nombre])

for nombre, edad in edades.items():
    print(f"{nombre} tiene {edad}")
```

### El patrón más útil: contar frecuencias

```python
texto = "abracadabra"
cuenta = {}
for letra in texto:
    if letra in cuenta:
        cuenta[letra] += 1
    else:
        cuenta[letra] = 1
print(cuenta)
```

Más corto con `.get`:

```python
cuenta = {}
for letra in "abracadabra":
    cuenta[letra] = cuenta.get(letra, 0) + 1
print(cuenta)
```

Y la versión de la librería estándar, `Counter`:

```python
from collections import Counter
print(Counter("abracadabra"))
print(Counter("abracadabra").most_common(1))   # el más frecuente
```

## ¿Cuál uso?

| Necesito... | Uso |
|---|---|
| Datos en orden, por posición | lista |
| Datos que no cambian (coordenadas) | tupla |
| Saber si algo está, sin repetidos | `set` |
| Asociar una clave con un valor, contar | `dict` |
