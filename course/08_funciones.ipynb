# Lección 8: Funciones

Una **función** es un pedazo de código con nombre que puedes usar muchas veces. Es como una receta guardada: la escribes una vez y la usas cuando quieras. 🧑‍🍳

Ya usaste funciones hechas por otros: `print`, `len`, `sum`, `int`. Ahora crearás las tuyas.

```python
def saludar():
    print("¡Hola!")
    print("Bienvenido a la escuela de programación")

saludar()
saludar()
```

- `def` crea la función.
- Lo que va con sangría es el cuerpo.
- Escribir `saludar()` la **llama**: ejecuta el cuerpo.

## Parámetros: darle datos a la función

```python
def saludar(nombre):
    print(f"Hola, {nombre}")

saludar("Ana")
saludar("Luis")
```

## `return`: devolver un resultado

`print` muestra algo en pantalla. `return` **entrega un valor** para que lo uses después.

```python
def doble(n):
    return n * 2

resultado = doble(21)
print(resultado)
print(doble(5) + doble(10))
```

Una función con varios parámetros y una decisión:

```python
def mayor(a, b):
    if a > b:
        return a
    return b

print(mayor(3, 9))
print(mayor(10, 2))
```

Una función puede devolver **varios valores** a la vez:

```python
def dividir(a, b):
    return a // b, a % b

cociente, residuo = dividir(17, 5)
print(cociente, residuo)
```

## Funciones que usan bucles

```python
def suma_hasta(n):
    total = 0
    for i in range(1, n + 1):
        total += i
    return total

print(suma_hasta(10))
print(suma_hasta(100))
```

```python
def es_primo(n):
    if n < 2:
        return False
    i = 2
    while i * i <= n:
        if n % i == 0:
            return False
        i += 1
    return True

print([x for x in range(30) if es_primo(x)])
```

💡 Para saber si `n` es primo solo hace falta probar divisores hasta la raíz cuadrada (`i * i <= n`). Esa idea ahorra muchísimo tiempo: la volveremos a ver.

## Variables dentro y fuera

Las variables creadas dentro de una función **solo existen ahí dentro**.

```python
def prueba():
    secreto = 42
    return secreto

print(prueba())
# print(secreto)   # daría error: 'secreto' no existe fuera de la función
```

## Valores por defecto

```python
def potencia(base, exponente=2):
    return base ** exponente

print(potencia(5))        # usa exponente=2
print(potencia(2, 10))
```

## Funciones que se llaman a sí mismas (recursión)

Una función puede llamarse a sí misma, siempre que tenga un **caso base** que detenga las llamadas.

```python
def factorial(n):
    if n <= 1:           # caso base
        return 1
    return n * factorial(n - 1)

print(factorial(5))
print(factorial(10))
```

## ✅ Resumen

- `def nombre(parametros):` crea una función; `nombre(valores)` la usa.
- `return` entrega un resultado; `print` solo lo muestra.
