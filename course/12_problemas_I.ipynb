Primero, ejecuta esta celda que simula la entrada (la viste en la lección 10):

```python
import io
import sys


def con_entrada(texto):
    """Simula el archivo de entrada: lo que escribas es lo que el programa 'leerá'."""
    sys.stdin = io.StringIO(texto.strip() + "\n")
    return sys.stdin.readline
```

## Problema 1: Suma de dos números

Te dan dos enteros `A` y `B`. Muestra su suma.

**Entrada:** una línea con `A` y `B`.
**Salida:** `A + B`.

| Entrada | Salida |
|---|---|
| `3 4` | `7` |
| `1000000000 1000000000` | `2000000000` |

💡 **Pista:** `a, b = map(int, input().split())`.

**Tu turno:**

```python
input = con_entrada("""
3 4
""")

# tu código aquí
```

**Solución:**

```python
input = con_entrada("""
3 4
""")

a, b = map(int, input().split())
print(a + b)
```

## Problema 2: Par o impar

Te dan un entero `N`. Muestra `PAR` si es par y `IMPAR` si es impar.

**Entrada:** un entero `N`.
**Salida:** `PAR` o `IMPAR` (en mayúsculas).

| Entrada | Salida |
|---|---|
| `10` | `PAR` |
| `7` | `IMPAR` |

💡 **Pista:** `n % 2`. Y respeta las mayúsculas: `PAR`, no `Par`.

**Tu turno:**

```python
input = con_entrada("""
7
""")

# tu código aquí
```

**Solución:**

```python
input = con_entrada("""
7
""")

n = int(input())
if n % 2 == 0:
    print("PAR")
else:
    print("IMPAR")
```

## Problema 3: El mayor de tres

Te dan tres enteros. Muestra el mayor.

**Entrada:** una línea con tres enteros.
**Salida:** el mayor.

| Entrada | Salida |
|---|---|
| `4 9 2` | `9` |
| `-5 -1 -8` | `-1` |

💡 **Pista:** hay una función de Python que ya lo hace. Pero también puedes resolverlo con `if`.

**Tu turno:**

```python
input = con_entrada("""
4 9 2
""")

# tu código aquí
```

**Solución:**

```python
input = con_entrada("""
-5 -1 -8
""")

a, b, c = map(int, input().split())
print(max(a, b, c))
```

## Problema 4: Suma de 1 hasta N

Te dan `N`. Muestra `1 + 2 + 3 + ... + N`.

**Entrada:** un entero `N` (`1 ≤ N ≤ 10^18`).
**Salida:** la suma.

| Entrada | Salida |
|---|---|
| `10` | `55` |
| `1000000000000` | `500000000000500000000000` |

💡 **Pista:** con `N` tan grande, un bucle tardaría años (lo viste en la lección 11). Recuerda la fórmula `N * (N + 1) // 2`.

**Tu turno:**

```python
input = con_entrada("""
10
""")

# tu código aquí
```

**Solución:**

```python
input = con_entrada("""
1000000000000
""")

n = int(input())
print(n * (n + 1) // 2)
```

## Problema 5: Contar positivos

Te dan `N` números. Cuenta cuántos son **mayores que cero**.

**Entrada:** primera línea `N`; segunda línea, los `N` números.
**Salida:** la cantidad de positivos.

| Entrada | Salida |
|---|---|
| `5`<br>`3 -1 0 7 -4` | `2` |

💡 **Pista:** lee la lista y usa un acumulador (lección 5).

**Tu turno:**

```python
input = con_entrada("""
5
3 -1 0 7 -4
""")

# tu código aquí
```

**Solución:**

```python
input = con_entrada("""
5
3 -1 0 7 -4
""")

n = int(input())
arr = [int(x) for x in input().split()]

contador = 0
for x in arr:
    if x > 0:
        contador += 1
print(contador)
```

## Problema 6: Promedio

Te dan `N` números. Muestra su promedio con **2 decimales**.

**Entrada:** primera línea `N`; segunda línea, los `N` números.
**Salida:** el promedio con exactamente 2 decimales.

| Entrada | Salida |
|---|---|
| `4`<br>`1 2 3 5` | `2.75` |
| `3`<br>`4 4 4` | `4.00` |

💡 **Pista:** `sum(arr) / n` da un decimal. Para mostrar exactamente 2 decimales usa un f-string: `f"{valor:.2f}"`.

**Tu turno:**

```python
input = con_entrada("""
4
1 2 3 5
""")

# tu código aquí
```

**Solución:**

```python
input = con_entrada("""
4
1 2 3 5
""")

n = int(input())
arr = [int(x) for x in input().split()]
promedio = sum(arr) / n
print(f"{promedio:.2f}")
```
