# Lección 4: Tomar decisiones con `if`

Hasta ahora el programa hace siempre lo mismo. Con `if` puede **decidir**. 🚦

> Si llueve, llevo paraguas. Si no, no.

```python
llueve = True

if llueve:
    print("Llevo paraguas")
```

## La regla de oro de Python: la sangría

Las líneas que van **dentro** del `if` se escriben con 4 espacios de sangría (o una tecla Tab). Esa sangría le dice a Python qué instrucciones pertenecen al `if`. Los `:` al final de la línea del `if` son obligatorios.

```python
edad = 15

if edad >= 12:
    print("Puedes entrar")
    print("Disfruta el juego")
print("Esto se muestra siempre")
```

## Comparaciones

| Operador | Significa |
|---|---|
| `==` | igual a |
| `!=` | distinto de |
| `<` | menor que |
| `>` | mayor que |
| `<=` | menor o igual |
| `>=` | mayor o igual |

⚠️ **Cuidado:** `=` guarda un valor, `==` compara. Confundirlos es el error más común.

```python
print(5 == 5)
print(5 != 5)
print(3 < 7)
print(3 >= 7)
```

## `else` y `elif`

- `else`: lo que pasa si **no** se cumple la condición.
- `elif`: otra condición para probar si la anterior falló.

```python
nota = 75

if nota >= 90:
    print("Excelente")
elif nota >= 70:
    print("Aprobado")
else:
    print("A estudiar un poco más")
```

Python revisa de arriba hacia abajo y entra solo en **el primer** bloque cuya condición sea verdadera.

## `and`, `or`, `not`

- `and`: las dos condiciones deben cumplirse.
- `or`: basta con una.
- `not`: invierte el resultado.

```python
edad = 13
tiene_permiso = True

if edad >= 12 and tiene_permiso:
    print("Puede ir de excursión")

dia = "sábado"
if dia == "sábado" or dia == "domingo":
    print("Hay fin de semana")

if not tiene_permiso:
    print("Falta el permiso")
```

Una condición con varias partes se puede escribir como un rango:

```python
x = 15
if 10 <= x <= 20:
    print("x está entre 10 y 20")
```

## Ejemplo: par o impar

```python
n = int(input("Escribe un número: "))

if n % 2 == 0:
    print("par")
else:
    print("impar")
```

- `if` / `elif` / `else` deciden qué bloque se ejecuta.

- `==` compara y `=` guarda.
