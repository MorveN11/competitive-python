Llegó el momento. 🎓 Ya sabes Python, sabes leer entradas y has resuelto problemas. Ahora aprenderás a usar el **template** de este repositorio: tu herramienta para competir.

El template hace tres cosas por ti:

1. Prepara la **entrada y salida** (local vs. juez).
2. Te da lectura **rápida** de datos.
3. Te regala un **debugger** (`dbg`) para ver qué pasa dentro de tu programa.


## El flujo de trabajo en 5 pasos

1. Crea un archivo `.py` en la raíz (por ejemplo `suma.py`).
2. Escribe `cp` y presiona **Tab**: aparece el template completo.
3. Pega la entrada de prueba en `data/input.txt`.
4. Presiona **Ctrl+Shift+B** para ejecutar.
5. Mira `data/output.txt` (tu respuesta) y `data/debug.txt` (tus mensajes).

### Parte 1: ¿local o juez?

```python
if os.path.exists(os.path.join(DATA, "input.txt")):
    sys.stdin = open(... "input.txt" ...)
    sys.stdout = open(... "output.txt" ...)
    sys.stderr = open(... "debug.txt" ...)
```

Un programa tiene tres "canales":

| Canal | Qué es | Normalmente va a | Con el template va a |
|---|---|---|---|
| `stdin` | lo que **entra** (`input()`) | teclado | `data/input.txt` |
| `stdout` | lo que **sale** (`print()`) | pantalla | `data/output.txt` |
| `stderr` | mensajes de **debug** | pantalla | `data/debug.txt` |

- En tu computador existe `data/input.txt`, así que los tres canales van a archivos.
- En el **juez online** ese archivo no existe, así que el código funciona normal con la consola.

**Tú no cambias nada** para enviar al juez: subes el mismo archivo.

### `dbg`: ver el valor de cualquier cosa

Muestra el nombre de la variable y su valor. Funciona con números, textos, listas, conjuntos, diccionarios...

````python
n = 5
arr = [4, 8, 15, 16, 23]
nombres = {"Ana", "Luis"}
edades = {"Ana": 11, "Luis": 12}

dbg(n)
dbg(arr)
dbg(nombres)
dbg(edades)
dbg(n, arr, label="inicio")      # varios a la vez, con una etiqueta
````

### `dbg_mat`: ver una matriz como tabla

````python
tabla = [[i * j for j in range(1, 6)] for i in range(1, 5)]
dbg_mat(tabla)
````

### `dbg_grid`: ver un tablero

````python
tablero = [
    "#..#.",
    ".#...",
    "...##",
]
dbg_grid(tablero)
````

### `dbg_sep` y `dbg_time`

- `dbg_sep("texto")` dibuja una línea para separar partes.
- `dbg_time("texto")` muestra cuánto tiempo pasó (útil para saber si tu programa es lento).

````python
dbg_sep("sumando")
total = 0
for i in range(1, 100001):
    total += i
dbg(total)
dbg_time("tras sumar")
````

### Cómo se usa `dbg` para encontrar un error

Este programa debería sumar solo los números pares, pero da mal resultado. Con `dbg` miramos qué pasa vuelta a vuelta:

````python
def suma_pares(arr):
    total = 0
    for x in arr:
        if x % 2 == 1:           # ¡hay un error aquí!
            total += x
        dbg(x, total, label="vuelta")
    return total

print(suma_pares([1, 2, 3, 4]))   # debería ser 6
````

Con `dbg` ves que `total` suma los números **impares**: el `if` debería ser `x % 2 == 0`. Esa es la forma más rápida de encontrar errores: **mira lo que pasa dentro**.

Corrige el programa:

````python
def suma_pares(arr):
    total = 0
    for x in arr:
        if x % 2 == 0:
            total += x
    return total

print(suma_pares([1, 2, 3, 4]))
````

## Un problema completo con el template

Resolvamos el problema **"Suma de la lista"** con todo lo aprendido.

> **Entrada:** primera línea `N`; segunda línea, `N` enteros.
> **Salida:** la suma de los números.

Así se ve la solución dentro del template (solo cambian las líneas dentro de `solve`):

```python
def solve():
    n = int(input())
    arr = [int(x) for x in input().split()]
    dbg(n, arr)
    ans = sum(arr)
    print(ans)
```

## Problemas comunes

| Síntoma | Qué revisar |
|---|---|
| `data/output.txt` está vacío | Mira el error en la terminal: algo falló antes del `print`. |
| `ValueError` al leer | `data/input.txt` no tiene el formato que espera tu código. |
| `EOFError` | Tu código intenta leer más líneas de las que hay en `data/input.txt`. |
| `IndexError` | Te saliste de los límites de una lista; revisa los índices. |
| `dbg` no aparece | Mira `data/debug.txt`, no `output.txt`. |
| `cp` + Tab no funciona | Mira la sección *Snippet* del README principal. |
