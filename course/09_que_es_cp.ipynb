# Lección 9: ¿Qué es la programación competitiva?

¡Ya sabes Python! Ahora vas a usarlo para algo emocionante: **resolver problemas como en una olimpiada**. 🏆

## La idea

La programación competitiva es un deporte mental. Te dan un **problema** escrito en palabras, y tienes que escribir un programa que lo resuelva. Un programa llamado **juez** lo prueba automáticamente con muchos ejemplos secretos y te dice si está bien.

Se parece a un acertijo, pero con tres reglas:
1. Tu programa debe dar la respuesta **correcta** en **todos** los casos.
2. Debe ser **rápido** (normalmente menos de 1 o 2 segundos).
3. Debe leer y escribir los datos exactamente con el **formato** que pide el problema.

## Las partes de un problema

Todos los problemas tienen la misma estructura. Ejemplo:

> **Suma de dos números**
>
> Te dan dos enteros `A` y `B`. Muestra su suma.
>
> **Entrada:** una línea con dos enteros `A` y `B` (`1 ≤ A, B ≤ 10^9`).
>
> **Salida:** un solo entero: `A + B`.
>
> | Entrada | Salida |
> |---|---|
> | `3 4` | `7` |

- **Enunciado:** la historia o lo que hay que calcular.
- **Entrada:** cómo vendrán los datos.
- **Salida:** cómo debes mostrar la respuesta.
- **Restricciones** (`1 ≤ A, B ≤ 10^9`): qué tan grandes pueden ser los datos. ¡Son muy importantes! Te dicen qué tan rápido debe ser tu solución.
- **Ejemplos:** casos de prueba con su respuesta correcta.

## El juez no perdona: los veredictos

Cuando envías tu código, el juez responde con una de estas siglas:

| Sigla | Significado | Qué hacer |
|---|---|---|
| **AC** | *Accepted*: ¡correcto! 🎉 | Celebrar y seguir con el siguiente |
| **WA** | *Wrong Answer*: respuesta incorrecta | Buscar casos que no pensaste |
| **TLE** | *Time Limit Exceeded*: muy lento | Buscar una idea más rápida |
| **RE** | *Runtime Error*: el programa se rompió | Revisar índices, división entre 0, formato de entrada |
| **CE** | *Compilation Error*: error de sintaxis | Revisar la escritura del código |

Casi nadie resuelve un problema a la primera. **Equivocarse es parte del juego.**

## Cómo se compite

- Hay un conjunto de problemas y un **tiempo límite** (por ejemplo, 3 horas).
- Ganas puntos por cada problema resuelto.
- Hay competencias individuales y por equipos.
- Existen olimpiadas escolares nacionales e internacionales (como la IOI) y plataformas para practicar en línea todo el año.

## Un primer problema, paso a paso

Problema: *dados dos números, mostrar su suma.*

Ya sabes resolverlo con lo que aprendiste. Esta es la versión con datos fijos:

```python
A = 3
B = 4
print(A + B)
```

Pero el juez **no** te da el 3 y el 4 escritos en tu código. Los entrega por la **entrada** y tu programa debe **leerlos**. Cómo leer esa entrada es el tema de la siguiente lección.

## Consejos de campeón 🥇

- **Lee el problema dos veces** antes de escribir código.
- **Prueba con los ejemplos** a mano antes de programar.
- Piensa en los **casos extremos**: el número más pequeño, el más grande, la lista vacía, todos iguales.
- Si te atoras, dibuja el problema en papel.
- Practica un poco **todos los días**: es mejor que mucho un solo día.
- Equivocarse enseña más que acertar.
