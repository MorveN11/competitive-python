## Simular la entrada en el cuaderno

En una competencia la entrada viene del teclado o de un archivo. Aquí usaremos esta pequeña función para **simular** esa entrada y probar nuestros programas. Ejecútala una vez al inicio.

````python
import io
import sys


def con_entrada(texto):
    """Simula el archivo de entrada: lo que escribas es lo que el programa 'leerá'."""
    sys.stdin = io.StringIO(texto.strip() + "\n")
    return sys.stdin.readline
````

Úsala así: `input = con_entrada("""...""")`. Desde ese momento `input()` lee de ese texto.

## Leer una línea

`input()` lee **una línea** completa y la entrega como **texto**.

````python
input = con_entrada("""
hola mundo
""")

linea = input()
print(repr(linea))      # repr muestra los caracteres escondidos
````

Ves que la línea termina con `\n` (salto de línea). Con `.strip()` lo quitamos. Cada `input()` lee la **siguiente** línea.

````python
input = con_entrada("""
primera
segunda
""")

print(input().strip())
print(input().strip())
````

> En el template de este repositorio, `input` es `sys.stdin.readline`: es lo mismo pero mucho más rápido. Con `readline` la línea siempre incluye el `\n`, así que `.strip()` o `int(...)` se encargan de limpiarlo. `int("5\n")` funciona bien.

## Receta 1: un entero

````python
input = con_entrada("""
42
""")

n = int(input())
print(n * 2)
````

## Receta 2: varios enteros en una línea

`.split()` parte la línea en pedazos separados por espacios. `map(int, ...)` convierte cada pedazo en entero.

````python
input = con_entrada("""
3 4
""")

a, b = map(int, input().split())
print(a + b)
````

Si no sabes cuántos son, guárdalos en una lista:

````python
input = con_entrada("""
5 1 9 2 7
""")

arr = [int(x) for x in input().split()]
print(arr)
print(sum(arr))
````

## Receta 3: la cantidad primero, luego los datos

Es el formato **más común**:

```
5
10 20 30 40 50
```

La primera línea dice cuántos datos vendrán; la segunda los trae.

````python
input = con_entrada("""
5
10 20 30 40 50
""")

n = int(input())
arr = [int(x) for x in input().split()]
print(n, arr)
print(max(arr))
````

## Receta 4: varias líneas, una por dato

````python
input = con_entrada("""
3
7
2
9
""")

n = int(input())
nums = [int(input()) for _ in range(n)]
print(nums)
````

## Receta 5: una matriz

```
3 4
1 2 3 4
5 6 7 8
9 0 1 2
```

````python
input = con_entrada("""
3 4
1 2 3 4
5 6 7 8
9 0 1 2
""")

filas, cols = map(int, input().split())
mat = [[int(x) for x in input().split()] for _ in range(filas)]
print(mat)
````

## Receta 6: texto

````python
input = con_entrada("""
Ana
programacion competitiva
""")

nombre = input().strip()
frase = input().strip()
print(nombre.upper())
print(frase.split())
````

Un tablero de caracteres (cada línea es una fila):

````python
input = con_entrada("""
3
#..
.#.
..#
""")

n = int(input())
grid = [input().strip() for _ in range(n)]
print(grid)
print(grid[1][1])
````

## Receta 7: varios casos de prueba

Muchos problemas traen primero `t` (la cantidad de casos) y luego `t` casos seguidos:

````python
input = con_entrada("""
3
2 5
10 20
7 7
""")

t = int(input())
for _ in range(t):
    a, b = map(int, input().split())
    print(a + b)
````

## Escribir la salida

`print` escribe una línea. Reglas de oro:

- Muestra **solo** lo que pide el problema: nada de "La respuesta es:".
- Respeta el formato exacto (espacios, mayúsculas, `YES` y no `Yes`).
- Para mostrar varios números en una línea, usa `join`:

````python
resultados = [3, 1, 4, 1, 5]

print(*resultados)                           # 3 1 4 1 5  (forma corta)
print(" ".join(map(str, resultados)))        # lo mismo, con join
print("\n".join(map(str, resultados)))       # uno por línea
````

## `input()` normal vs rápido

| | `input()` normal | `sys.stdin.readline` |
|---|---|---|
| Velocidad | lenta con mucha entrada | muy rápida |
| Incluye `\n` | no | sí |
| En el template | no se usa | sí: `input = sys.stdin.readline` |

Con pocos datos da igual. Con 10^5 líneas o más, la diferencia puede ser la diferencia entre **AC** y **TLE**. Por eso nuestro template ya viene con la versión rápida.

## ✅ Resumen

- `int(input())` → un entero.
- `a, b = map(int, input().split())` → varios enteros en una línea.
