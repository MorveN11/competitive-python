# Lección 11: Cómo pensar un problema y qué es "rápido"

Saber Python no basta: hay que saber **pensar**. Esta lección te da un método y una idea clave: la velocidad.

## El método de 6 pasos 🧭

1. **Entender:** lee el problema dos veces. ¿Qué datos me dan? ¿Qué me piden?
2. **Ejemplos:** resuelve a mano los ejemplos. ¿Obtienes la misma respuesta?
3. **Idea:** piensa cómo lo hiciste a mano. Eso es tu algoritmo.
4. **Casos extremos:** ¿qué pasa con el menor, el mayor, todos iguales, uno solo?
5. **Código:** ahora sí, escribe el programa.
6. **Probar:** ejecuta los ejemplos y tus propios casos extremos.

Un **algoritmo** es simplemente una receta paso a paso para resolver un problema.

## Ejemplo completo: el número mayor

> Te dan `n` números. Muestra el mayor.
>
> Entrada: `5` y luego `3 8 2 9 4`. Salida: `9`.

**Idea (a mano):** voy mirando los números; me quedo con el mayor que he visto hasta ahora. Si veo uno más grande, lo reemplazo.

```python
numeros = [3, 8, 2, 9, 4]

mayor = numeros[0]
for x in numeros:
    if x > mayor:
        mayor = x
print(mayor)
```

**Casos extremos:** ¿un solo número? ¿todos negativos? Por eso empezamos con `mayor = numeros[0]` y no con `0`: si todos fueran negativos, `0` daría una respuesta incorrecta.

```python
print(max([5]))           # un solo número
print(max([-7, -3, -9]))  # todos negativos
```

## ¿Qué es "rápido"?

Python hace unas **10 a 30 millones** de vueltas de bucle por segundo. Esa cifra manda: si tu programa necesita muchas más que eso, recibirás **TLE**. ⏱️

Veamos una carrera. Problema: sumar los números de 1 a `n`.

**Solución lenta:** un bucle que suma uno por uno (`n` operaciones).
**Solución rápida:** una fórmula, sin bucle (1 operación).

```python
import time

n = 3_000_000

t0 = time.perf_counter()
suma = 0
for i in range(1, n + 1):
    suma += i
t_lenta = time.perf_counter() - t0

t0 = time.perf_counter()
formula = n * (n + 1) // 2
t_rapida = time.perf_counter() - t0

print("Mismo resultado:", suma == formula)
print(f"Bucle:   {t_lenta:.4f} segundos")
print(f"Fórmula: {t_rapida:.8f} segundos")
```

Las dos dan la misma respuesta, pero una es **millones de veces más rápida**. Con `n = 10^12` el bucle no terminaría nunca y la fórmula sigue siendo instantánea.

## Contar las operaciones

Para saber si tu idea es suficientemente rápida:

1. Mira las **restricciones** del problema (cuánto vale `N` como máximo).
2. Cuenta cuántas veces se repite tu código en el peor caso.

| Tu código | Operaciones (peor caso) | Notación |
|---|---|---|
| un `for` de `n` vueltas | `n` | O(n) |
| un `for` dentro de otro `for` | `n × n` | O(n²) |
| una fórmula | 1 | O(1) |
| dividir entre 2 hasta llegar a 1 | `log₂ n` (≈30 para mil millones) | O(log n) |

Esa notación con la O se llama **complejidad**. Guía rápida para Python con límite de 1 segundo:

| Máximo de `N` | Complejidad que sirve |
|---|---|
| hasta ~10 | casi cualquier cosa |
| hasta ~1 000 | O(n²) |
| hasta ~100 000 | O(n) u O(n log n) |
| hasta ~10^18 | O(log n) o una fórmula |

Ejemplo: si `N ≤ 100 000`, un bucle doble haría 10^10 operaciones: **TLE seguro**. Hay que encontrar otra idea.

## El truco que más TLE evita: usar `set` en lugar de lista para buscar

Preguntar `x in lista` revisa uno por uno. Preguntar `x in conjunto` es casi instantáneo.

```python
import time

datos = list(range(5000))
conjunto = set(datos)
buscar = list(range(5000, 10000))         # ninguno está: peor caso

t0 = time.perf_counter()
encontrados = sum(1 for x in buscar if x in datos)
t_lista = time.perf_counter() - t0

t0 = time.perf_counter()
encontrados = sum(1 for x in buscar if x in conjunto)
t_set = time.perf_counter() - t0

print(f"Lista: {t_lista:.4f} segundos")
print(f"Set:   {t_set:.6f} segundos")
```

## Ejemplo: encontrar duplicados rápido

> ¿Hay algún número repetido en la lista?

**Lenta (O(n²)):** comparar cada par.
**Rápida (O(n)):** ir guardando en un `set` lo que ya vi.

```python
def hay_repetido(lista):
    vistos = set()
    for x in lista:
        if x in vistos:
            return True
        vistos.add(x)
    return False

print(hay_repetido([3, 1, 4, 1, 5]))
print(hay_repetido([3, 1, 4, 2, 5]))
```

## Probar tus ideas: casos de prueba propios

Antes de enviar al juez, inventa tus propios casos pequeños **que sepas la respuesta**:

```python
def mayor(lista):
    m = lista[0]
    for x in lista:
        if x > m:
            m = x
    return m

# Pruebas: assert falla con error si la condición es falsa
assert mayor([3, 8, 2, 9, 4]) == 9
assert mayor([5]) == 5
assert mayor([-7, -3, -9]) == -3
print("Todas las pruebas pasaron ✅")
```

`assert` es una forma sencilla de comprobar que tu código hace lo que esperas.
