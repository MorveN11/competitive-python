# Lección 5: Repetir con bucles

Imagina escribir "Hola" 1000 veces a mano. 😱 Los **bucles** le piden a la computadora que repita instrucciones por nosotros. Son la herramienta más importante de la programación competitiva.

## `for` con `range`: repetir un número de veces

```python
for i in range(5):
    print("Hola")
```

`range(5)` produce los números `0, 1, 2, 3, 4`. **Empieza en 0 y termina antes del 5.** La variable `i` toma cada valor, uno por vuelta:

```python
for i in range(5):
    print("vuelta número", i)
```

### Variantes de `range`

| Escribes | Produce |
|---|---|
| `range(5)` | 0, 1, 2, 3, 4 |
| `range(2, 6)` | 2, 3, 4, 5 |
| `range(0, 10, 2)` | 0, 2, 4, 6, 8 (de 2 en 2) |
| `range(5, 0, -1)` | 5, 4, 3, 2, 1 (hacia atrás) |

```python
for i in range(1, 6):
    print(i)

print("---")
for i in range(10, 0, -2):
    print(i)
```

Atención: el final nunca se incluye. Para contar del 1 al `n` inclusive usa `range(1, n + 1)`.

## Acumuladores: la técnica estrella

Una variable que **va juntando** un resultado mientras el bucle avanza. Ejemplo: sumar del 1 al 10.

```python
suma = 0                      # empieza en cero
for i in range(1, 11):
    suma += i                 # en cada vuelta, súmale i
print(suma)
```

Así se ve por dentro, vuelta por vuelta:

```python
suma = 0
for i in range(1, 6):
    suma += i
    print(f"i = {i}  ->  suma = {suma}")
```

Otro ejemplo: contar cuántos números del 1 al 100 son múltiplos de 7.

```python
contador = 0
for i in range(1, 101):
    if i % 7 == 0:
        contador += 1
print(contador)
```

## `while`: repetir mientras algo sea verdad

Se usa cuando no sabes cuántas vueltas habrá.

```python
n = 100
pasos = 0
while n > 1:
    n = n // 2
    pasos += 1
print("Dividí entre 2", pasos, "veces para llegar a 1")
```

⚠️ Si la condición nunca deja de ser verdadera, el bucle **no termina nunca** (bucle infinito). Asegúrate de que algo dentro cambie la condición.

## `break` y `continue`

- `break`: sale del bucle de inmediato.
- `continue`: salta a la siguiente vuelta.

```python
for i in range(1, 20):
    if i % 2 == 0:
        continue          # salta los pares
    if i > 11:
        break             # termina el bucle
    print(i)
```

## Bucles dentro de bucles

Un bucle dentro de otro sirve para tablas, cuadrículas y matrices.

```python
for fila in range(1, 4):
    for col in range(1, 5):
        print(fila * col, end=" ")    # end=" " evita el salto de línea
    print()                           # salto de línea al terminar la fila
```

Un triángulo de estrellas:

```python
n = 5
for i in range(1, n + 1):
    print("*" * i)
```

## ✅ Resumen

- `for i in range(n)` repite `n` veces, con `i` de 0 a `n-1`.
- `range(a, b, paso)`: empieza en `a`, termina antes de `b`.
- Un acumulador junta un resultado vuelta a vuelta.
- `while` repite mientras la condición sea verdadera.
- `break` sale del bucle; `continue` salta a la siguiente vuelta.
