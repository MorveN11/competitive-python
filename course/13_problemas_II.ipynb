Ejecuta primero la función que simula la entrada:

```python
import io
import sys


def con_entrada(texto):
    """Simula el archivo de entrada: lo que escribas es lo que el programa 'leerá'."""
    sys.stdin = io.StringIO(texto.strip() + "\n")
    return sys.stdin.readline
```

## Problema 1: Palíndromo

Una palabra es **palíndromo** si se lee igual al derecho y al revés. Te dan una palabra. Muestra `SI` si es palíndromo y `NO` si no.

**Entrada:** una palabra en minúsculas.
**Salida:** `SI` o `NO`.

| Entrada | Salida |
|---|---|
| `reconocer` | `SI` |
| `python` | `NO` |

💡 **Pista:** `palabra[::-1]` la da vuelta.

**Tu turno:**

```python
input = con_entrada("""
reconocer
""")

# tu código aquí
```

**Solución:**

```python
input = con_entrada("""
reconocer
""")

palabra = input().strip()
print("SI" if palabra == palabra[::-1] else "NO")
```

## Problema 2: Letra más frecuente

Te dan una palabra. Muestra la letra que más veces aparece. Si hay empate, muestra la menor en orden alfabético.

**Entrada:** una palabra en minúsculas.
**Salida:** la letra y cuántas veces aparece, separadas por un espacio.

| Entrada | Salida |
|---|---|
| `abracadabra` | `a 5` |
| `zzaa` | `a 2` |

💡 **Pista:** cuenta con un diccionario (lección 7). Recorre las letras **ordenadas** para que, en caso de empate, gane la menor.

**Tu turno:**

```python
input = con_entrada("""
abracadabra
""")

# tu código aquí
```

**Solución:**

```python
input = con_entrada("""
zzaa
""")

palabra = input().strip()

cuenta = {}
for letra in palabra:
    cuenta[letra] = cuenta.get(letra, 0) + 1

mejor = None
for letra in sorted(cuenta):                  # en orden alfabético
    if mejor is None or cuenta[letra] > cuenta[mejor]:
        mejor = letra
print(mejor, cuenta[mejor])
```

## Problema 4: Suma de cada fila

Te dan una matriz de `F` filas y `C` columnas. Muestra la suma de cada fila, una por línea.

**Entrada:** primera línea `F` y `C`; luego `F` líneas con `C` números.
**Salida:** `F` líneas, la suma de cada fila.

| Entrada | Salida |
|---|---|
| `3 3`<br>`1 2 3`<br>`4 5 6`<br>`7 8 9` | `6`<br>`15`<br>`24` |

💡 **Pista:** lee la matriz como en la lección 10 y aplica `sum` a cada fila.

**Tu turno:**

```python
input = con_entrada("""
3 3
1 2 3
4 5 6
7 8 9
""")

# tu código aquí
```

**Solución:**

```python
input = con_entrada("""
3 3
1 2 3
4 5 6
7 8 9
""")

filas, cols = map(int, input().split())
mat = [[int(x) for x in input().split()] for _ in range(filas)]

for fila in mat:
    print(sum(fila))
```

## Problema 5: Pasos de Collatz

Empieza con un número `N`. Repite estas reglas hasta llegar a 1:
- Si `N` es par, divídelo entre 2.
- Si `N` es impar, multiplícalo por 3 y súmale 1.

Muestra **cuántos pasos** necesitas para llegar a 1.

**Entrada:** un entero `N` (`1 ≤ N ≤ 10^6`).
**Salida:** el número de pasos.

| Entrada | Salida | Explicación |
|---|---|---|
| `6` | `8` | 6 → 3 → 10 → 5 → 16 → 8 → 4 → 2 → 1 |
| `1` | `0` | ya está en 1 |

💡 **Pista:** un `while n != 1` con un contador de pasos.

**Tu turno:**

```python
input = con_entrada("""
6
""")

# tu código aquí
```

**Solución:**

```python
input = con_entrada("""
6
""")

n = int(input())
pasos = 0
while n != 1:
    if n % 2 == 0:
        n //= 2
    else:
        n = 3 * n + 1
    pasos += 1
print(pasos)
```

## Problema 6: Varios casos de prueba

Te dan `T` casos. En cada uno hay un número `N`: muestra `PRIMO` si es primo y `NO` si no lo es.

**Entrada:** primera línea `T`; luego `T` líneas, cada una con un entero `N` (`1 ≤ N ≤ 10^9`).
**Salida:** una línea por caso.

| Entrada | Salida |
|---|---|
| `4`<br>`7`<br>`10`<br>`1`<br>`97` | `PRIMO`<br>`NO`<br>`NO`<br>`PRIMO` |

💡 **Pista:** reutiliza la función `es_primo` de la lección 8. Con `N` hasta 10^9, solo hace falta probar divisores hasta la raíz cuadrada (unos 31 000).

**Tu turno:**

```python
input = con_entrada("""
4
7
10
1
97
""")

# tu código aquí
```

**Solución:**

```python
input = con_entrada("""
4
7
10
1
97
""")


def es_primo(n):
    if n < 2:
        return False
    i = 2
    while i * i <= n:
        if n % i == 0:
            return False
        i += 1
    return True


t = int(input())
for _ in range(t):
    n = int(input())
    print("PRIMO" if es_primo(n) else "NO")
```
