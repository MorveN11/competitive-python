## `print`: hacer que la computadora hable

La instrucción `print` muestra algo en pantalla. Lo que quieres mostrar va entre paréntesis.

```python
print("¡Hola, mundo!")
```

El texto va entre comillas `"..."`. A un texto se le llama **string** (cadena).

Puedes usar `print` muchas veces. La computadora ejecuta las instrucciones **de arriba hacia abajo**.

```python
print("Hola")
print("Me llamo Python")
print("Hoy vamos a aprender a programar")
```

Los números no necesitan comillas. Python incluso puede hacer cuentas:

```python
print(7)
print(2 + 3)
print(10 * 4)
```

Fíjate en la diferencia entre estas dos:

```python
print(2 + 3)      # una cuenta: muestra 5
print("2 + 3")    # un texto: muestra 2 + 3
```

## Comentarios

Todo lo que va después de un `#` lo ignora la computadora. Sirve para dejar notas para humanos, como ya viste arriba.

```python
# Esto es un comentario, Python no lo lee
print("Esto sí se ejecuta")  # también puedes comentar al final de una línea
```

## Los errores son normales 🐞

Los programadores se equivocan **todo el tiempo**, incluso los mejores del mundo. Cuando algo está mal escrito, Python te avisa con un mensaje de error. Ese mensaje es tu amigo: te dice qué línea falló.

Algunos errores típicos (no los ejecutes todavía, solo mira):

| Código | Qué pasa | Arreglo |
|---|---|---|
| `print("Hola)` | falta cerrar las comillas | `print("Hola")` |
| `Print("Hola")` | Python distingue mayúsculas de minúsculas | `print("Hola")` |
| `print "Hola"` | faltan los paréntesis | `print("Hola")` |

Regla de oro: **lee el error, mira la línea que menciona, y busca la letra o símbolo que falta**.
