## Qué significa "con grader"

En un problema normal tu programa lee con `input()` y escribe con `print()`. En un problema **con grader** no: el juez ya trae un programa (el *grader*) que **lee la entrada, llama a tu función y muestra el resultado**. Tú solo escribes la función con el **nombre y los parámetros exactos** que pide el enunciado.

```python
# Lo que escribes tú: SOLO la función.
def contar_pares(arr):
    return sum(1 for x in arr if x % 2 == 0)


# Lo que hace el juez (grader): leer, llamar a tu función y mostrar.
def grader():
    arr = [3, 8, 2, 9, 4]
    print(contar_pares(arr))


grader()
```

### Reglas de oro con grader

1. **Nombre y parámetros exactos** de la función, tal como los pide el enunciado.
2. **No leas** la entrada ni imprimas nada extra (`print` de depuración incluido): puede romper la comunicación con el grader. Si necesitas depurar, usa `dbg` (va a `stderr`, no afecta la salida).
3. **Devuelve** el resultado con `return`; no lo imprimas, salvo que el enunciado lo pida.
4. **No dependas de variables globales** que se arrastren entre llamadas: el grader puede llamar tu función varias veces (un caso por llamada).
5. **Cuidado con modificar los argumentos**: si recibes una lista y la ordenas con `.sort()`, cambias la lista del grader. Si dudas, trabaja sobre una copia: `sorted(arr)` o `arr[:]`.
6. Respeta los **tipos** que pide: si debe devolver un entero, no devuelvas `3.0`.

## Python para quien viene de C++

Diferencias que causan los errores más comunes:

```python
# 1. División entera y módulo con negativos: Python redondea hacia ABAJO
assert -7 // 2 == -4          # C++ daría -3
assert -7 % 3 == 2            # C++ daría -1; en Python el residuo es >= 0 con módulo positivo

# 2. Los enteros no se desbordan
assert 2 ** 100 > 10 ** 30

# 3. El operador / siempre da decimal; usa // para enteros
assert 7 / 2 == 3.5 and 7 // 2 == 3

# 4. Cuidado con los alias: asignar una lista NO la copia
a = [1, 2, 3]
b = a
b.append(4)
assert a == [1, 2, 3, 4]

c = a[:]                      # esta sí es una copia (superficial)
c.append(5)
assert a == [1, 2, 3, 4]

# 5. La trampa de las matrices
mal = [[0] * 3] * 2           # las dos filas son LA MISMA lista
mal[0][0] = 1
assert mal[1][0] == 1

bien = [[0] * 3 for _ in range(2)]
bien[0][0] = 1
assert bien[1][0] == 0
print("ok")
```

## Complejidad: cuántas operaciones caben

CPython ejecuta del orden de **10^7 vueltas de bucle simples por segundo** (3.11 es algo más rápido). Con el límite de tiempo del problema y el tamaño máximo `N`, elige un algoritmo:

| `N` máximo | Complejidad que alcanza en Python |
|---|---|
| ≤ 10 | `O(n!)`, `O(n · 2^n)` |
| ≤ 20 | `O(2^n)` |
| ≤ 300 | `O(n³)` |
| ≤ 3 000 | `O(n²)` (con cuidado) |
| ≤ 10^5 – 10^6 | `O(n)` o `O(n log n)` |
| ≤ 10^9 – 10^18 | `O(log n)`, `O(√n)` si `n ≤ 10^12` o fórmula |

**Trucos para ir más rápido en Python:**

- Usa funciones internas: `sum`, `min`, `max`, `sorted`, `any`, `all`, `math.gcd`. Corren en C.
- Prefiere comprensiones de listas a bucles con `append`.
- Busca en `set`/`dict` (`O(1)`), no en listas (`O(n)`).
- Para colas usa `collections.deque`; `list.pop(0)` es `O(n)`.
- Junta cadenas con `"".join(lista)`, no con `s += ...` en un bucle.
- Evita recursión profunda (límite por defecto ≈ 1000). Si hace falta: `sys.setrecursionlimit(10**6)`.
