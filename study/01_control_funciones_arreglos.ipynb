## Repaso rápido de control

```python
# if / elif / else
def clasificar(n):
    if n < 0:
        return "negativo"
    elif n == 0:
        return "cero"
    return "positivo"


assert clasificar(-3) == "negativo" and clasificar(0) == "cero" and clasificar(8) == "positivo"

# for con range, while, break, continue
suma = 0
for i in range(1, 11):
    if i % 2 == 0:
        continue
    suma += i
assert suma == 25          # 1+3+5+7+9

# for ... else: el else se ejecuta si el for NO hizo break
def es_primo(n):
    if n < 2:
        return False
    for d in range(2, int(n ** 0.5) + 1):
        if n % d == 0:
            break
    else:
        return True
    return False


assert [x for x in range(10) if es_primo(x)] == [2, 3, 5, 7]
```

## Funciones

- Una función recibe parámetros y **devuelve** un valor con `return`.
- Puede devolver varios valores (una tupla).
- Las variables de dentro no existen fuera.
- Un parámetro puede tener valor por defecto.

```python
def minmax(arr):
    return min(arr), max(arr)


lo, hi = minmax([4, 9, 1, 7])
assert (lo, hi) == (1, 9)


def potencia(base, exp=2):
    return base ** exp


assert potencia(5) == 25 and potencia(2, 10) == 1024
```

### Recursión

Una función que se llama a sí misma con un problema **más pequeño**. Siempre necesita un **caso base**.

```python
def factorial(n):
    if n <= 1:
        return 1
    return n * factorial(n - 1)


def suma_digitos(n):
    if n < 10:
        return n
    return n % 10 + suma_digitos(n // 10)


assert factorial(5) == 120
assert suma_digitos(4729) == 22
```

## Arreglos (listas) 1D

| Operación | Costo | Nota |
|---|---|---|
| `a[i]` | O(1) | índices desde 0; negativos cuentan desde el final |
| `a.append(x)` | O(1) | agrega al final |
| `a.pop()` | O(1) | quita el último |
| `a.insert(0, x)`, `a.pop(0)` | **O(n)** | lento: usa `deque` si lo necesitas |
| `x in a` | **O(n)** | usa `set` para búsquedas |
| `a.sort()` / `sorted(a)` | O(n log n) | `sort` modifica; `sorted` copia |
| `a[i:j]` | O(j-i) | crea una copia |

```python
a = [5, 2, 8, 1]

assert a[0] == 5 and a[-1] == 1
assert a[1:3] == [2, 8]
assert a[::-1] == [1, 8, 2, 5]
assert sorted(a) == [1, 2, 5, 8] and a == [5, 2, 8, 1]      # sorted no modifica a

# Patrones frecuentes
cuenta = [0] * 5                      # contador por índice
for x in [1, 3, 3, 4, 1, 1]:
    cuenta[x] += 1
assert cuenta == [0, 3, 0, 2, 1]

pares = [x for x in a if x % 2 == 0]  # filtrar
dobles = [x * 2 for x in a]           # transformar
assert pares == [2, 8] and dobles == [10, 4, 16, 2]

# enumerate: índice y valor a la vez
assert [(i, v) for i, v in enumerate("ab")] == [(0, "a"), (1, "b")]

# Intercambiar dos elementos
a[0], a[3] = a[3], a[0]
assert a == [1, 2, 8, 5]
```

## Arreglos multidimensionales (matrices)

Una matriz es una **lista de listas**: `m[fila][columna]`.

```python
n, m_ = 3, 4
mat = [[0] * m_ for _ in range(n)]       # n filas, m columnas (¡no uses [[0]*m]*n!)
mat[1][2] = 7
assert mat == [[0, 0, 0, 0], [0, 0, 7, 0], [0, 0, 0, 0]]

m = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9],
]
filas, cols = len(m), len(m[0])

# Recorrido por filas
total = 0
for i in range(filas):
    for j in range(cols):
        total += m[i][j]
assert total == 45

# Suma por fila y por columna
suma_filas = [sum(f) for f in m]
suma_cols = [sum(m[i][j] for i in range(filas)) for j in range(cols)]
assert suma_filas == [6, 15, 24] and suma_cols == [12, 15, 18]

# Diagonal principal y secundaria (matriz cuadrada)
diag1 = [m[i][i] for i in range(filas)]
diag2 = [m[i][filas - 1 - i] for i in range(filas)]
assert diag1 == [1, 5, 9] and diag2 == [3, 5, 7]

# Transpuesta y giro de 90° (sentido horario)
trasp = [list(f) for f in zip(*m)]
giro = [list(f) for f in zip(*m[::-1])]
assert trasp == [[1, 4, 7], [2, 5, 8], [3, 6, 9]]
assert giro == [[7, 4, 1], [8, 5, 2], [9, 6, 3]]
```

### Vecinos en una cuadrícula

Casi todos los problemas de matrices necesitan mirar las celdas de al lado. Usa una lista de **direcciones** y valida que no te salgas.

```python
DIR4 = [(-1, 0), (1, 0), (0, -1), (0, 1)]                              # arriba, abajo, izq, der
DIR8 = DIR4 + [(-1, -1), (-1, 1), (1, -1), (1, 1)]                     # con diagonales


def vecinos(i, j, filas, cols, dirs=DIR4):
    for di, dj in dirs:
        ni, nj = i + di, j + dj
        if 0 <= ni < filas and 0 <= nj < cols:      # dentro de la matriz
            yield ni, nj


assert sorted(vecinos(0, 0, 3, 3)) == [(0, 1), (1, 0)]            # esquina: 2 vecinos
assert len(list(vecinos(1, 1, 3, 3))) == 4                        # centro: 4 vecinos
assert len(list(vecinos(1, 1, 3, 3, DIR8))) == 8
```

### Matrices 3D

Una lista de matrices: `cubo[capa][fila][col]`.

```python
capas, filas, cols = 2, 3, 4
cubo = [[[0] * cols for _ in range(filas)] for _ in range(capas)]
cubo[1][2][3] = 9
assert cubo[1][2][3] == 9 and cubo[0][2][3] == 0
```

## Problemas tipo (estilo grader)

**P1.** `suma_bordes(m)`: suma de los elementos del borde de una matriz.
**P2.** `celdas_con_vecino_igual(m)`: cuántas celdas tienen algún vecino (4 direcciones) con el mismo valor.
**P3.** `rotar(m)`: giro de 90° horario.
**P4.** `espiral(m)`: lista de los elementos recorridos en espiral.

```python
def suma_bordes(m):
    f, c = len(m), len(m[0])
    total = 0
    for i in range(f):
        for j in range(c):
            if i in (0, f - 1) or j in (0, c - 1):
                total += m[i][j]
    return total


def celdas_con_vecino_igual(m):
    f, c = len(m), len(m[0])
    cuenta = 0
    for i in range(f):
        for j in range(c):
            for ni, nj in vecinos(i, j, f, c):
                if m[ni][nj] == m[i][j]:
                    cuenta += 1
                    break                     # basta con uno
    return cuenta


def rotar(m):
    return [list(fila) for fila in zip(*m[::-1])]


def espiral(m):
    res = []
    arriba, abajo, izq, der = 0, len(m) - 1, 0, len(m[0]) - 1
    while arriba <= abajo and izq <= der:
        for j in range(izq, der + 1):
            res.append(m[arriba][j])
        for i in range(arriba + 1, abajo + 1):
            res.append(m[i][der])
        if arriba < abajo:
            for j in range(der - 1, izq - 1, -1):
                res.append(m[abajo][j])
        if izq < der:
            for i in range(abajo - 1, arriba, -1):
                res.append(m[i][izq])
        arriba += 1
        abajo -= 1
        izq += 1
        der -= 1
    return res


m = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
assert suma_bordes(m) == 40                       # todo menos el 5
assert celdas_con_vecino_igual([[1, 1, 2], [3, 4, 2]]) == 4
assert rotar([[1, 2], [3, 4]]) == [[3, 1], [4, 2]]
assert espiral(m) == [1, 2, 3, 6, 9, 8, 7, 4, 5]
assert espiral([[1, 2, 3, 4]]) == [1, 2, 3, 4]
print("ok")
```
